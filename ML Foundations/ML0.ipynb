# ML Foundations

The models, the evaluation discipline, and the data handling that everything after this
assumes. Deep learning (Part IV) is a *specialization* of these ideas, not a replacement for
them — and in production, the models in this part still do more real work than neural networks
do.

## Coming from a data science role? Skim, don't skip.

If you've been doing applied DS or experimentation work, most of this is already yours. That's
genuinely an advantage — plenty of MLE candidates are weaker here than you are.

**Self-check.** If you can answer all of these comfortably, skim ML1–ML6 and read ML7
properly:

1. Why does L1 regularization produce exact zeros when L2 doesn't?
2. Why must you fit a scaler on the training split only — and what specifically leaks if you don't?
3. What does bagging reduce, what does boosting reduce, and why does that difference matter?
4. When does a high ROC-AUC coexist with a useless model?
5. Why do tree-based models not need feature scaling?

If several of those are shaky, read the chapter in question in full.

**What is probably new, regardless of your background** — and where this part is worth your
time even if the algorithms aren't:

- **ML7** as a whole. You know precision and recall. What's likely new is choosing an operating
  threshold against a business constraint, calibration as a distinct property from ranking
  quality, and connecting an offline metric to an online one you can actually move.
- The **"What changes in production"** section closing each chapter. This is the transition
  material: a `StandardScaler` in a notebook is a line of code, but in production it is a
  versioned artifact that must be reproduced *exactly* at serving time or your model silently
  degrades. The algorithms don't change when you deploy them. Everything around them does.

## How this part connects

- **Part V (ML Coding interviews)** re-implements much of this from scratch in NumPy. ML1 → mlc1,
  ML5/ML6 → mlc2, ML7 → mlc5. Learn it here, implement it there.
- **Part IV (Deep Learning)** builds directly on ML1's loss/gradient-descent material and ML2's
  overfitting material. DL3 is ML2's regularization section specialized to neural networks.
- **Part VII (ML System Design)** assumes all of it. `mlsd3`'s "GBDT vs deep models" is
  unreadable without ML5.

## Key Takeaways
- These models are still the production default for tabular problems — gradient-boosted trees
  remain the thing to beat, and often aren't beaten
- Evaluation discipline (ML2) and leakage avoidance (ML3) transfer unchanged to deep learning and LLMs
- If you're strong here already, spend your time on ML7 and the production-delta sections
